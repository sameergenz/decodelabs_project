Visualization(graphs)

```python
import matplotlib.pyplot as plt
df["Product"].value_counts().plot(kind="bar")
plt.title("Product Count")
plt.xlabel("Product")
plt.ylabel("Count")
plt.show()
```

```python
#payment method Bar Chart
df["PaymentMethod"].value_counts().plot(kind="bar")
plt.title("Payment Mathod Count")
plt.xlabel("Payment Mathod")
plt.ylabel("Count")
plt.show()
```

```python
#orderStatus pie chart
df["OrderStatus"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Order Status Distribution")
plt.ylabel("")
plt.show()
```

```python
df["TotalPrice"].plot(kind="hist",bins=20)
plt.title("Total Price Distribution")
plt.xlabel("Total Price")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.boxplot(df["TotalPrice"])

plt.title("Box Plot of Total Price")
plt.ylabel("Total Price")

plt.show()
```

OBSERVATION
1. Most orders have a Total Price below ₹1000.
2. The number of orders decreases as Total Price increases.
3. The distribution is positively (right) skewed.
4. Very high-value orders are relatively rare.
